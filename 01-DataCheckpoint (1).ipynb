## Research Question

Research question: How does the amount of sleep college students get each night relate to their academic performance, and can sleep patterns help predict changes in GPA or course grades?


Independent Variable (Sleep Amount): Average number of hours a college student sleeps per night, measured through self-reported survey responses over a typical school week.

Independent Variable (Sleep Patterns): Consistency of sleep schedule, including average bedtime, wake-up time, and variation in sleep duration between weekdays and weekends.

Dependent Variable (Academic Performance): Student GPA and/or percentage course grades, measured using self-reported cumulative GPA or grades from current courses.

Population: College students enrolled in undergraduate courses.

## Background and Prior Work

Sleep plays an important role in cognitive functioning, learning, memory, and academic success. College students often struggle with inconsistent sleep schedules due to academic pressure, social activities, work responsibilities, and technology use. These irregular sleep habits can negatively affect concentration, motivation, and overall classroom performance. Because of this, researchers have spent years studying the relationship between sleep and academic achievement among college students.

Previous research has consistently shown a strong connection between sleep quality and academic performance. In a study conducted by Gilbert and Weaver, researchers surveyed university students about their sleep habits and compared the results to their grade point averages (GPA). The study found that students with poorer sleep quality and greater daytime sleepiness tended to have lower GPAs (Gilbert & Weaver, 2010). Their findings suggested that not only the amount of sleep, but also the quality of sleep, plays an important role in academic success. However, the study relied heavily on self-reported data, which may reduce accuracy because students may incorrectly estimate their sleep patterns or academic behaviors.
Similarly, Curcio, Ferrara, and De Gennaro (2006) conducted a large review of previous studies examining the effects of sleep deprivation on learning and academic performance. Their review analyzed evidence from multiple experiments and concluded that sleep loss negatively impacts attention span, memory consolidation, problem-solving skills, and the ability to learn new information. The researchers explained that students who consistently lose sleep may struggle to retain information learned in class, which can lead to lower academic performance over time (Curcio et al., 2006). One limitation of this review is that many of the studies included different age groups and research methods, making it difficult to generalize the results specifically to college students.

In addition, Hershner and Chervin (2014) examined the causes and consequences of sleep deprivation among college students. Their research reviewed common factors that contribute to poor sleep, including stress, heavy coursework, electronic device use, and irregular schedules. The study found that insufficient sleep was linked to reduced alertness, impaired learning, poorer academic performance, and increased mental health concerns among college students (Hershner & Chervin, 2014). The researchers also emphasized that many students underestimate the importance of sleep and fail to recognize how sleep deprivation affects their daily functioning. While the study provided a broad overview of sleep-related issues, it focused more on general consequences of sleep deprivation rather than predicting specific changes in GPA or course grades.
Overall, past research strongly supports the idea that sleep affects how well students perform academically. Prior studies show that poor sleep quality and sleep deprivation are associated with lower GPA, reduced learning ability, and weaker academic outcomes. However, many studies rely on self-reported sleep data or focus broadly on academic performance rather than predicting specific changes in GPA or course grades over time. These limitations help support the purpose of this project, which aims to further examine whether sleep patterns can help predict academic performance among college students.

References
Gilbert, S. P., & Weaver, C. C. (2010). Sleep quality and academic performance in university students. College Student Journal, 44(2), 295–303.

Hershner, S., & Chervin, R. (2014). Causes and consequences of sleepiness among college students. Nature and Science of Sleep, 6, 73–84.

### CMU Sleep and Psychology Dataset 

The primary dataset used in this project is the CMU Sleep and Psychology Dataset, available through the CMU Statistics Data Repository. This dataset contains information related to sleep behaviors, stress, caffeine consumption, and measures of cognitive or academic functioning. The dataset includes over 600 participant observations and multiple variables that capture different aspects of sleep and psychological well-being. Each row in the dataset represents an individual participant, while each column represents a measured variable, making the dataset relatively tidy and straightforward to analyze.

Several variables in the dataset are especially important for this project. Sleep Duration measures the number of hours participants sleep each night and is useful for examining whether insufficient sleep is associated with reduced cognitive performance or increased fatigue. Sleep Quality provides a subjective measure of how restful participants perceive their sleep to be, which may be just as important as total sleep time. The dataset also includes variables related to Stress Levels, which can influence both sleep habits and academic functioning, and Caffeine Consumption, which may temporarily improve alertness while also disrupting long-term sleep quality. Additional variables measuring daytime tiredness, fatigue, or performance outcomes help provide insight into how sleep behaviors affect daily functioning.

This dataset has several limitations that are important to consider. Most of the information is self-reported, meaning participants may unintentionally misreport their sleep habits, stress levels, or caffeine intake. The sample may also be biased because participants were likely students or volunteers interested in psychology or health research, which may not accurately represent the broader population. In addition, sleep patterns among college students are often influenced by irregular schedules, academic pressure, and social activities, making the findings less generalizable to non-student groups. The dataset may also contain missing values or outliers, such as unusually low sleep durations or extremely high caffeine consumption, which will need to be examined and cleaned during preprocessing.
